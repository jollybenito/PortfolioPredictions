```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import yfinance as yf
from tqdm import tqdm
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
import plotly
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import plotly.express as px
import plotly.figure_factory as ffz
import glob

path = "stock_market_data/sp500/csv"
# csv files in the path
files = glob.glob(path + "/*.csv")

# Calculate lags
def lagit(df, lags):
    for i in range(1, lags + 1):
        df['Lag_'+str(i)] = df['return'].shift(i)
    return ['Lag_'+str(i) for i in range(1, lags + 1)]

# Define starter scenario which is the exception case for the loop
i=0
# defining an empty list to store 
data_frame_Price = pd.DataFrame()
data_frame_Lag1 = pd.DataFrame()
data_frame_Lag2 = pd.DataFrame()

content = []
# checking all the csv files in the 
# specified path
for filename in files:
    # reading content of csv file
    # content.append(filename)
    df = pd.read_csv(filename, index_col=None)
    df['Date'] = pd.to_datetime(df['Date'], format="%d-%m-%Y")
    df.sort_values(by='Date', ascending=True, inplace=True)
    # We are going to focus on recent predictions so I will limit the timeframe
    # to years above 2000 
    df = df[df.Date>'2005-01-01']

    df = df[["Date", "Close"]]
    df['return'] = df['Close'].pct_change()
    df['direction'] = np.where(df['return']>0, 1, 0)
    features = lagit(df, 3)
    df.dropna(inplace=True)

    assetname = filename.split("/")[-1].split(".csv")[0]
    df_Price = df[["Date","return"]].copy()
    df_Price.columns = ["Date",assetname]
    df_Lag1 = df[["Date","Lag_1"]].copy()
    df_Lag1.columns = ["Date",assetname]
    df_Lag2 = df[["Date","Lag_2"]].copy()
    df_Lag2.columns = ["Date",assetname]
    content.append(df_Price)
    if (i == 0):
        data_frame_Price = df_Price
        data_frame_Lag1 = df_Lag1
        data_frame_Lag2 = df_Lag2
        i = 1
    else:
        data_frame_Price = pd.merge(data_frame_Price, df_Price,
                how="outer", on="Date")
        data_frame_Lag1 = pd.merge(data_frame_Lag1, df_Lag1,
                how="outer", on="Date")
        data_frame_Lag2 = pd.merge(data_frame_Lag2, df_Lag2,
                how="outer", on="Date")
```

```python
import numpy as np
import pandas as pd
import yfinance as yf
from scipy.optimize import minimize

def calculate_portfolio_variance(weights, cov_matrix):
    return np.dot(weights.T, np.dot(cov_matrix, weights))

def calculate_risk_contribution(weights, cov_matrix):
    portfolio_variance = calculate_portfolio_variance(weights, cov_matrix)
    marginal_contrib = np.dot(cov_matrix, weights)
    risk_contrib = np.multiply(weights, marginal_contrib) / portfolio_variance
    return risk_contrib

def risk_parity_objective(weights, cov_matrix):
    risk_contrib = calculate_risk_contribution(weights, cov_matrix)
    target_risk = np.mean(risk_contrib)
    return np.sum((risk_contrib - target_risk) ** 2)

def get_risk_parity_weights(cov_matrix):
    num_assets = len(cov_matrix)
    initial_weights = np.ones(num_assets) / num_assets
    constraints = ({'type': 'eq', 'fun': lambda x: np.sum(x) - 1})
    bounds = [(0, 1) for _ in range(num_assets)]
    
    result = minimize(risk_parity_objective, initial_weights, args=(cov_matrix,), 
                      method='SLSQP', bounds=bounds, constraints=constraints)
    
    return result.x

# Covariance 
cov_matrix = return_vec.cov()
# Annualized covariance, 252 trading days for a year
cov_matrix = return_vec.cov()*252
# There is a minimum weight size since there is a minimum fraction to buy of stock
risk_parity_weights = get_risk_parity_weights(cov_matrix)
print("Risk Parity Weights:", risk_parity_weights)
```

```python
import pandas as pd
import numpy as np
from skfolio.optimization import MeanRisk, HierarchicalRiskParity, NestedClustersOptimization
from skfolio.prior import EmpiricalPrior
from skfolio.moments import EquilibriumMu, ShrunkMu, DenoiseCovariance, LedoitWolf, OAS
from sklearn.model_selection import train_test_split
import cvxpy as cp
cp.installed_solvers()

# Split data into training/testing sets
X_train, X_test = train_test_split(return_vec, test_size=0.4, shuffle=False)
# Define the prior estimator (used for covariance matrix estimation)
prior_estimator = EmpiricalPrior(mu_estimator=ShrunkMu(), covariance_estimator=OAS())
#prior_estimator = EmpiricalPrior()

# 4. Define the optimization model
# 'Mean Risk' 
model_MR = MeanRisk(min_weights=0.0, max_weights=1.0, budget=1.0,
    cardinality=300, prior_estimator=prior_estimator, risk_free_rate=0.05,
    solver="SCIP" # A specialized solver like Mosek is often needed for MIP
)
# 5. Fit the model to the data (optimization occurs here)
model_MR.fit(X_train)

# 6. Get the results
portfolio_performance = model_MR.predict(X_test)
nonzero_assets = X_test[portfolio_performance.nonzero_assets]
nonzero_weights = portfolio_performance.weights[portfolio_performance.weights>0]
mean_risk_returns = np.dot(nonzero_assets, nonzero_weights) #(nonzero_assets * nonzero_weights).sum().sum()

# 'HierarchicalRiskParity' aims to equalize the risk contributions of selected assets
model_HRP = HierarchicalRiskParity(min_weights=0.0, max_weights=1.0, 
    prior_estimator=prior_estimator, 
)
# 5. Fit the model to the data (optimization occurs here)
model_HRP.fit(X_train)

# 6. Get the results
portfolio_performance = model_HRP.predict(X_test)
HRP_nonzero_assets = X_test[portfolio_performance.nonzero_assets]
HRP_nonzero_weights = portfolio_performance.weights[portfolio_performance.weights>0]
hierarchical_risk_returns = np.dot(HRP_nonzero_assets, HRP_nonzero_weights) #(nonzero_assets * nonzero_weights).sum().sum()
```
